# Data Collection and Initial Inspection

## Objectives

This notebook:

- documents the location and provenance of the project data;
- confirms that the required files are available;
- examines dataset dimensions and column names;
- investigates missing values and duplicated records;
- identifies the available date range;
- records initial data-quality observations.

No source data is modified in this notebook.

```python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = (
    Path.cwd().parent
    if Path.cwd().name == "jupyter_notebooks"
    else Path.cwd()
)

RAW_FOLDER = PROJECT_ROOT / "data" / "raw" / "v1"

GLOBAL_FILE = RAW_FOLDER / "GlobalTemperatures.csv"
COUNTRY_FILE = RAW_FOLDER / "GlobalLandTemperaturesByCountry.csv"

print(f"Project root: {PROJECT_ROOT}")
print(f"Global file: {GLOBAL_FILE}")
print(f"Country file: {COUNTRY_FILE}")
```

```python
assert GLOBAL_FILE.exists(), (
    f"Missing file: {GLOBAL_FILE}. "
    "Download it from Kaggle and place it in data/raw/v1."
)

assert COUNTRY_FILE.exists(), (
    f"Missing file: {COUNTRY_FILE}. "
    "Download it from Kaggle and place it in data/raw/v1."
)

print("Both required dataset files were found.")
```

```python
global_df = pd.read_csv(GLOBAL_FILE)
country_df = pd.read_csv(COUNTRY_FILE)

print(f"Global dataset shape: {global_df.shape}")
print(f"Country dataset shape: {country_df.shape}")
```

```python
display(global_df.head())
display(country_df.head())
```

```python
expected_global_columns = {
    "dt",
    "LandAverageTemperature",
    "LandAverageTemperatureUncertainty",
    "LandMaxTemperature",
    "LandMaxTemperatureUncertainty",
    "LandMinTemperature",
    "LandMinTemperatureUncertainty",
    "LandAndOceanAverageTemperature",
    "LandAndOceanAverageTemperatureUncertainty",
}

expected_country_columns = {
    "dt",
    "AverageTemperature",
    "AverageTemperatureUncertainty",
    "Country",
}

missing_global_columns = expected_global_columns.difference(
    global_df.columns
)

missing_country_columns = expected_country_columns.difference(
    country_df.columns
)

assert not missing_global_columns, (
    f"Missing expected global columns: {missing_global_columns}"
)

assert not missing_country_columns, (
    f"Missing expected country columns: {missing_country_columns}"
)

print("All expected columns are present.")
```

```python
print("GLOBAL DATASET")
global_df.info()

print("\nCOUNTRY DATASET")
country_df.info()
```

```python
def missing_value_summary(dataframe):
    """Return the number and percentage of missing values per column."""
    summary = pd.DataFrame({
        "missing_count": dataframe.isna().sum(),
        "missing_percentage": (
            dataframe.isna().mean() * 100
        ).round(2),
    })

    return summary.sort_values(
        by="missing_percentage",
        ascending=False,
    )


print("Global dataset missing values")
display(missing_value_summary(global_df))

print("Country dataset missing values")
display(missing_value_summary(country_df))
```

```python
global_duplicate_dates = global_df.duplicated(
    subset=["dt"]
).sum()

country_duplicate_records = country_df.duplicated(
    subset=["Country", "dt"]
).sum()

print(f"Duplicated global dates: {global_duplicate_dates}")
print(
    "Duplicated country-and-date combinations: "
    f"{country_duplicate_records}"
)
```

```python
global_dates = pd.to_datetime(
    global_df["dt"],
    errors="coerce",
)

country_dates = pd.to_datetime(
    country_df["dt"],
    errors="coerce",
)

print("Global dataset")
print(f"Earliest date: {global_dates.min()}")
print(f"Latest date: {global_dates.max()}")
print(f"Invalid dates: {global_dates.isna().sum()}")

print("\nCountry dataset")
print(f"Earliest date: {country_dates.min()}")
print(f"Latest date: {country_dates.max()}")
print(f"Invalid dates: {country_dates.isna().sum()}")
```

```python
country_counts = (
    country_df.groupby("Country")
    .size()
    .sort_values(ascending=False)
)

print(f"Number of country labels: {country_df['Country'].nunique()}")

display(country_counts.head(10))
display(country_counts.tail(10))
```

## Initial Data-Quality Findings

- The global dataset contains **3,192 rows** and **9 columns**.
- The country dataset contains **577,462 rows** and **4 columns**.
- The global dataset covers **1750-01-01** to **2015-12-01**.
- The country dataset covers **1743-11-01** to **2013-09-01**.
- The most important missing-value findings were:
  **1,200 values (37.59%) are missing from each of the global maximum,
  minimum, and land-and-ocean temperature columns and their corresponding
  uncertainty columns. These variables contain no observations before
  January 1850. The global land-average temperature and uncertainty columns
  each contain 12 missing values (0.38%). In the country dataset,
  AverageTemperature contains 32,651 missing values (5.65%), while
  AverageTemperatureUncertainty contains 31,912 missing values (5.53%).
  Neither dataset contains missing dates, and the Country column contains
  no missing values.**
- The duplicated-record checks found:
  **0 duplicated dates in the global dataset and 0 duplicated
  country-and-date combinations in the country dataset.**
- The country field contains **243 distinct labels**.

These findings will be used to design the cleaning strategy. Missing
temperatures will not be filled automatically until their location and
frequency have been investigated.
